```python
import cv2
import pytesseract
from google.colab.patches import cv2_imshow
```

```python
image=cv2.imread('/content/Image20260108193508.png')
cv2_imshow(image)
```

```python
grey=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
cv2_imshow(grey)
```

```python
text=pytesseract.image_to_string(grey)
print(text)
```

```python
if "Margherita Pizza" or "Margharita Pizza" in text:
    print(" 'Margharita Pizza' found in the menu.")
else:
    print(" 'Margharita Pizza' not found in the menu.")
```
